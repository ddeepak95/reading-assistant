```python
from book_summarizer.core.highlight_processor import KindleHighlightParser
from pathlib import Path

def save_highlights_to_markdown(html_path: str, output_md_path: str):
    """
    Extract Kindle highlights from HTML file and save to markdown file using existing parser.
    
    Args:
        html_path: Path to the Kindle highlights HTML file
        output_md_path: Path where the markdown file should be saved
    """
    # Use the existing parser
    parser = KindleHighlightParser(html_path)
    
    # Create output directory if it doesn't exist
    output_path = Path(output_md_path)
    output_path.parent.mkdir(parents=True, exist_ok=True)
    
    # Write to markdown file
    with open(output_md_path, 'w', encoding='utf-8') as f:
        for chapter in parser.highlights:
            f.write(f"\n## {chapter['title']}\n\n")
            for highlight in chapter['highlights']:
                f.write(f"  > [!quote] {highlight}\n") 

# Example usage
if __name__ == "__main__":
    html_path = "./books/covenant-of-water/highlights.html"
    output_md_path = "./storage/notebook/Book Notes/The Covenant of Water.md"
    save_highlights_to_markdown(html_path, output_md_path)
```
