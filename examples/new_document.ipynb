```python
from book_summarizer import (
    DocumentHandler, 
    HighlightProcessor, 
    SummaryGenerator, 
    Config
)
from tinydb import TinyDB
import os, json

# Initialize configuration
Config.initialize()

# Initialize databases
main_db = TinyDB(Config.MAIN_DB_PATH)
highlights_db = TinyDB(Config.HIGHLIGHTS_DB_PATH)

# Initialize components
doc_handler = DocumentHandler(main_db, highlights_db)
highlight_processor = HighlightProcessor(highlights_db)
summary_generator = SummaryGenerator(openai_api_key="your-openai-api-key")
```

## Load a New Document

Now let's load a new document into the system. You can load just the document first and add highlights later.

```python
# Document paths
doc_name = "The Invisible Empire"
doc_source = "./books/invisible-empire/book.epub"

# Load document without highlights
doc_id = doc_handler.load_document(
    doc_name=doc_name,
    doc_source=doc_source
)

print(f"Document loaded with ID: {doc_id}")
```

```python
doc_id = "the-invisible-empire"

highlights_source = "../books/invisible-empire/highlights.html"

# check current os path
print(os.path.abspath(highlights_source))

# check if highlights exist
if os.path.exists(highlights_source):
    print(f"Highlights file found at: {highlights_source}")
else:
    print(f"Highlights file not found at: {highlights_source}")


# Process and add highlights
highlight_processor.process_highlights(
    highlights_path=highlights_source,
    doc_id=doc_id
)

```
